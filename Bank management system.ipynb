```python
#Bank  Account Management System

    class Bank:
        bankname="SBI"
        bankbranch="Peelamedu"
        bankloc="Coimbatore"
        bankIFSC="SBIN057689"
    
        def __init__(self,name,phno,address,accno,bal):
            self.name=name
            self.phno=phno
            self.address=address
            self.accno=accno
            self.bal=bal
    
        @classmethod
        def disp(cls):
            print("Bank Name:",cls.bankname)
            print("Bank Branch:",cls.bankbranch)
            print("Bank Location:",cls.bankloc)
            print("Bank IFSC:",cls.bankIFSC)
    
        @classmethod
        def change_bankloc(cls,new_bankloc):
            cls.bankloc=new_bankloc
            print("Bank Location Changed Successfully")
    
        @classmethod
        def change_bankbranch(cls,new_bankbranch):
            cls.bankbranch=new_bankbranch
            print("Bank Branch Changed Successfully")    
    
        @staticmethod
        def add(a,b):
            return a+b
    
        @staticmethod
        def sub(a,b):
            return a-b
    
        def display(self):
            print(self.name,self.phno,self.address,self.accno)
    
        def check_balance(self):
            print(f'{self.name} your current balance is RS.{self.bal}')
    
        def deposit(self,new_amount):
            self.bal=self.add(self.bal,new_amount)
            print(f"RS.{new_amount} deposited successfully")
    
        def withdraw(self,amount):
            if amount<=self.bal:
                self.bal=self.sub(self.bal,amount)
                print(f"{self.name} you withdrawned RS.{amount}")
                
            else:
                print("Insuffient balance,Kindly check your account balance")
    
        def change_phno(self,new_phno):
            self.phno=new_phno
            print(" Your phone number changed Successfully")
                
        def change_address(self,new_address):
            self.address=new_address
            print(" Your Address changed Successfully")
    
    customer1=Bank("Gayathri",987654325,"Srinivaasan Nagar,2nd Street,Puthu Road ,Tirupur",2335374774824536,35000)
    customer2=Bank("Gopika",9345775835,"Vaasan Nagar,3nd Street,Kovil vazhi ,Tirupur",23334353742628,23000)
    customer3=Bank("Ganga",9875474822,"Sri Nagar,6nd Street,thuyar Road ,Coimbatore",23353747352829,5000)
    
    
    customer1.display()
    print("--"*30)
    customer2.display()
    print("--"*30)
    customer2.check_balance()
    print("--"*30)
    customer1.deposit(30000)
    print("--"*30)
    customer1.check_balance()
    print("--"*30)
    Bank.disp()
    print("--"*30)
    customer3.disp()
    print("--"*30)
    Bank.change_bankloc("Kovai")
    print("--"*30)
    Bank.disp()
    print("--"*30)
    Bank.change_bankbranch("Sulur")
    print("--"*30)
    Bank.disp()
    print("--"*30)
    customer2.withdraw(2000)
    print("--"*30)
    customer2.check_balance()
    print("--"*30)
    customer3.withdraw(30000)
    print("--"*30)
```
